```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# Grid configuration for 105 images
nrows = 7
ncols = 15
n_images = 105

# Create the figure with your requested size
fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(15, 6))

# Flatten the 2D axes array (7x15) into a 1D list of 105 elements for easy looping
axes = axes.flatten()

# Example: Assuming your images are named 0.jpg to 104.jpg in a folder
for i in range(n_images):
    img_path = f"/data/add_disk0/liu47/flood_thesis_data/tufi_flood/TUFI_dataset/global_perspective_imgs/{1635+i}.jpg" 
    
    if os.path.exists(img_path):
        img = mpimg.imread(img_path)
        axes[i].imshow(img)
    else:
        # Fallback if an image is missing
        axes[i].text(0.5, 0.5, 'N/A', ha='center', va='center')
    axes[i].set_title(f"2_00{i+1}")
    
    # Remove borders and tick marks so the grid looks clean
    axes[i].axis("off")

# Removes excess whitespace between the 105 images
plt.tight_layout(pad=0.5)
plt.show()
```
